```python
# ===== NOTEBOOK 3 - STEP 2: load everything + summary table (mean ± std over 5 folds) =====
import os, glob, pickle
import numpy as np, pandas as pd
import matplotlib.pyplot as plt

meta = pd.read_csv(glob.glob("/kaggle/input/**/metadata_final.csv", recursive=True)[0])

# ---- 1) Load results + predictions of each task (if available) ----
R, P = {}, {}
for task in ["bin", "multi"]:
    f = glob.glob(f"/kaggle/input/**/results_{task}.csv", recursive=True)
    if not f:
        print(f"[{task}] no results yet")
        continue
    R[task] = pd.read_csv(f[0])
    with open(os.path.join(os.path.dirname(f[0]), f"predictions_{task}.pkl"), "rb") as fh:
        P[task] = pickle.load(fh)
    n_ok = (R[task]["precision_mode"] != "failed").sum()
    print(f"[{task}] {f[0]} -> {n_ok}/{len(R[task])} trainings OK")

MODELS  = ["VGG16", "ResNet50", "DenseNet121", "EfficientNetB0", "MobileNetV2"]
METRICS = ["auc", "balanced_acc", "f1", "sensitivity", "accuracy"]
TITLE   = {"bin": "Binary (Benign vs Malignant)", "multi": "Multiclass (Normal / Benign / Malignant)"}

# ---- 2) Table: mean ± std over the 5 folds, for each model and augmentation ----
def summary(task):
    r = R[task][R[task]["precision_mode"] != "failed"]
    g = r.groupby(["model", "augment"])
    out = pd.DataFrame(index=g.size().index)
    for m in METRICS:
        out[m] = g[m].mean().map("{:.3f}".format) + " ± " + g[m].std().map("{:.3f}".format)
    out["n_folds"]      = g.size()
    out["params_M"]     = g["params_M"].first().round(1)
    out["train_min"]    = (g["train_time_s"].mean() / 60).round(1)
    out["infer_ms/img"] = g["infer_ms_per_img"].mean().round(1)
    return out.reindex(MODELS, level=0)

for task in R:
    print(f"\n===== {TITLE[task]} =====")
    s = summary(task)
    display(s)
    s.to_csv(f"/kaggle/working/summary_{task}.csv")

# ---- 3) Chart: AUC per model, with vs without augmentation ----
fig, axes = plt.subplots(1, len(R), figsize=(7 * len(R), 4.5), squeeze=False)
for ax, task in zip(axes[0], R):
    r = R[task][R[task]["precision_mode"] != "failed"]
    s = r.groupby(["model", "augment"])["auc"].agg(["mean", "std"])
    x = np.arange(len(MODELS))
    for k, aug in enumerate([False, True]):
        d = s.xs(aug, level="augment").reindex(MODELS)
        ax.bar(x + (k - 0.5) * 0.38, d["mean"], 0.38, yerr=d["std"], capsize=3,
               label="with augmentation" if aug else "without augmentation")
    ax.set_xticks(x); ax.set_xticklabels(MODELS, rotation=20)
    ax.set_ylim(0.5, 1.0); ax.set_ylabel("AUC (mean ± std, 5 folds)")
    ax.set_title(TITLE[task]); ax.legend(); ax.grid(axis="y", alpha=0.3)
plt.tight_layout(); plt.savefig("/kaggle/working/fig_auc.png", dpi=150); plt.show()
```

```python
# ===== NOTEBOOK 3 - STEP 3: confusion matrices (5 folds together) + results per dataset =====
from sklearn.metrics import confusion_matrix, roc_auc_score

CLASSES = {"bin": ["Benign", "Malignant"], "multi": ["Normal", "Benign", "Malignant"]}

def oof(task, model, aug):
    """Test predictions of the 5 folds together (every image is tested exactly once)."""
    rows, yt, yp, pred = [], [], [], []
    for fold in range(5):
        p = P[task].get((model, aug, fold))
        if p is None or len(p["y_true"]) == 0:
            continue
        rows.append(p["rows"]); yt.append(p["y_true"]); yp.append(p["y_prob"])
        if task == "bin":     # binary: threshold chosen on validation (saved in the results)
            thr = R[task].query("model == @model and augment == @aug and fold == @fold")["threshold"].iloc[0]
            pred.append((p["y_prob"][:, 1] >= thr).astype(int))
        else:
            pred.append(p["y_prob"].argmax(1))
    return np.concatenate(rows), np.concatenate(yt), np.concatenate(yp), np.concatenate(pred)

# ---- 1) Confusion matrices: count + % of each true class ----
for task in R:
    names = CLASSES[task]
    fig, axes = plt.subplots(1, 5, figsize=(22, 4.2))
    for ax, model in zip(axes, MODELS):
        _, yt, _, pred = oof(task, model, False)
        cm  = confusion_matrix(yt, pred, labels=range(len(names)))
        cmn = cm / cm.sum(1, keepdims=True)
        ax.imshow(cmn, cmap="Blues", vmin=0, vmax=1)
        for i in range(len(names)):
            for j in range(len(names)):
                ax.text(j, i, f"{cm[i, j]}\n{100 * cmn[i, j]:.0f}%", ha="center", va="center",
                        color="white" if cmn[i, j] > 0.5 else "black", fontsize=9)
        ax.set_xticks(range(len(names))); ax.set_xticklabels(names, rotation=20)
        ax.set_yticks(range(len(names))); ax.set_yticklabels(names)
        ax.set_xlabel("Predicted"); ax.set_ylabel("True"); ax.set_title(model)
    plt.suptitle(f"{TITLE[task]} - confusion matrices (5 test folds together, without augmentation)")
    plt.tight_layout(); plt.savefig(f"/kaggle/working/fig_cm_{task}.png", dpi=150); plt.show()

# ---- 2) Shortcut check: results PER DATASET ----
# If the model only recognised the dataset, the AUC inside ONE dataset would fall to about 0.5
for task in R:
    last, out = len(CLASSES[task]) - 1, []
    for model in MODELS:
        rows, yt, yp, pred = oof(task, model, False)
        ds = meta.loc[rows, "dataset"].values
        for d in ["MIAS", "INbreast", "CBIS-DDSM"]:
            m = ds == d
            out.append({"model": model, "dataset": d, "n_images": int(m.sum()),
                        "accuracy": (pred[m] == yt[m]).mean(),
                        "AUC Malignant vs rest": roc_auc_score(yt[m] == last, yp[m][:, last])})
    t = (pd.DataFrame(out).pivot(index="model", columns="dataset",
                                 values=["n_images", "accuracy", "AUC Malignant vs rest"])
         .reindex(MODELS).round(3))
    print(f"\n===== {TITLE[task]}: results PER DATASET (without augmentation) =====")
    display(t)
    t.to_csv(f"/kaggle/working/per_dataset_{task}.csv")
```

```python
# ===== NOTEBOOK 3 - STEP 4: training curves + performance vs cost =====
EPOCHS_HEAD = 5     # phase 1 (CNN frozen) = 5 epochs, then fine-tuning

# ---- 1) Training curves: loss (train / validation) and validation AUC, for each fold ----
for task in R:
    fig, axes = plt.subplots(2, 5, figsize=(24, 8), sharey="row")
    for c, model in enumerate(MODELS):
        for fold in range(5):
            h = P[task].get((model, False, fold), {}).get("history", {})
            if not h:
                continue
            ep = np.arange(1, len(h["loss"]) + 1)
            axes[0, c].plot(ep, h["loss"], color=f"C{fold}", alpha=0.8)
            axes[0, c].plot(ep, h["val_loss"], color=f"C{fold}", ls="--", alpha=0.8)
            axes[1, c].plot(ep, h["val_auc_true"], color=f"C{fold}", label=f"fold {fold}")
        for r_ in range(2):
            axes[r_, c].axvline(EPOCHS_HEAD + 0.5, color="grey", ls=":")
            axes[r_, c].grid(alpha=0.3)
        axes[0, c].set_title(model); axes[1, c].set_xlabel("epoch")
    axes[0, 0].set_ylabel("loss (solid = train, dashed = validation)")
    axes[1, 0].set_ylabel("validation AUC"); axes[1, 0].legend(fontsize=8)
    plt.suptitle(f"{TITLE[task]} - training curves (without augmentation). Dotted line = start of fine-tuning")
    plt.tight_layout(); plt.savefig(f"/kaggle/working/fig_curves_{task}.png", dpi=150); plt.show()

# ---- 2) Performance vs cost (number of parameters, training time) ----
for task in R:
    r = R[task][(R[task]["precision_mode"] != "failed") & (~R[task]["augment"])]
    s = (r.groupby("model")
          .agg(auc=("auc", "mean"), auc_std=("auc", "std"), params=("params_M", "first"),
               train_min=("train_time_s", lambda x: x.mean() / 60))
          .reindex(MODELS))
    fig, axes = plt.subplots(1, 2, figsize=(13, 4.5))
    for ax, col, lab in [(axes[0], "params", "parameters (millions)"),
                         (axes[1], "train_min", "training time per fold (minutes)")]:
        ax.errorbar(s[col], s["auc"], yerr=s["auc_std"], fmt="o", capsize=3)
        for name, row in s.iterrows():
            ax.annotate(name, (row[col], row["auc"]), textcoords="offset points", xytext=(6, 6))
        ax.set_xlabel(lab); ax.set_ylabel("AUC (mean ± std)"); ax.grid(alpha=0.3)
    plt.suptitle(f"{TITLE[task]} - performance vs cost (without augmentation)")
    plt.tight_layout(); plt.savefig(f"/kaggle/working/fig_cost_{task}.png", dpi=150); plt.show()
```
